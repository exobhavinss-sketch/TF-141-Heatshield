###  Quick reference: methods you may need
These are either new or only briefly seen in class. Everything else you need was used in Labs 11-15.

| Method | What it does | Tiny example |
|:--|:--|:--|
| `df.duplicated()` / `df.drop_duplicates()` | Flags / removes rows identical to an earlier row (`keep=False` marks *all* copies) | `df.duplicated().sum()` |
| `s.value_counts()` | Counts each distinct value (`dropna=False` also counts NaN; `normalize=True` gives proportions) | `df["gender"].value_counts()` |
| `df.sort_values(by, ascending=...)` | Sorts rows by one or more columns | `df.sort_values(["age","bmi"], ascending=[False, True])` |
| `s.str.strip()`, `s.str.lower()` | Cleans text in a column | `df["gender"].str.strip().str.lower()` |
| `df.copy()` | Makes an independent copy | `df_clean = df.copy()` |
| `df.reset_index(drop=True)` | Renumbers the row labels 0, 1, 2, ... | `df.reset_index(drop=True)` |
| `np.where(cond, a, b)` | Picks `a` where the condition is True, else `b` (can be nested) | `np.where(df["age"] > 60, "old", "young")` |
| `s.isin([...])` | True where the value is one of the listed values | `df["work_type"].isin(["Private","Govt_job"])` |
| `df.isnull().any(axis=1)` | True for rows that contain at least one NaN | `df[df.isnull().any(axis=1)]` |
| `s.unique()` / `.tolist()` | The distinct values, as an array / list | `df["gender"].unique().tolist()` |
| `s.round(n)` | Rounds to `n` decimals | `(x * 100).round(2)` |
